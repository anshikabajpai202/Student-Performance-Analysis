# Student Performance Analysis

## Project Overview

This project focuses on analyzing student performance data to understand the factors that may influence students' academic scores.

The analysis aims to identify patterns in student performance, examine differences across various student groups, identify students who may be at risk of poor performance, and provide actionable recommendations that can help improve academic outcomes.

## Problem Statement

As a Data Analyst hired by a school, the objective is to analyze student performance data and provide useful insights to the school principal.

The analysis will focus on:
- Understanding the factors related to student scores
- Comparing performance across different student groups
- Identifying students who may be at risk
- Providing data-driven recommendations to improve student outcomes

## Dataset

The project uses the **Students Performance in Exams** dataset from Kaggle.

The dataset contains 600+ student records with information related to student characteristics, parental education, test preparation, and scores in mathematics, reading, and writing.

## Objectives

The main objectives of this project are:

1. Explore and clean the student performance dataset.
2. Analyze whether parental education level affects student scores.
3. Compare scores of students who completed the test preparation course with those who did not.
4. Analyze the correlation between reading, writing, and mathematics scores.
5. Compare performance between genders across different subjects.
6. Analyze the distribution of total student scores.
7. Identify and segment at-risk students.
8. Develop data-driven recommendations for improving student outcomes.

## Key Analysis Areas

The project will include:

- Data Exploration and Cleaning
- Factor Analysis
- Exploratory Data Analysis (EDA)
- Data Visualization
- At-Risk Student Segmentation
- Principal's Report
- Actionable Recommendations

## Tools & Technologies

- Python
- Pandas

- Matplotlib
- Seaborn

## Expected Visualizations

The project will include at least six visualizations:

1. Box plot – Scores by parental education
2. Bar chart – Test preparation comparison
3. Correlation heatmap – Subject scores
4. Grouped bar chart – Gender vs. subject performance
5. Histogram – Total score distribution
6. Scatter plot – Reading score vs. Mathematics score

## Project Outcome

The final analysis will provide data-backed findings about student performance, identify groups with higher risk of poor performance, and suggest practical actions that the school can implement to improve academic outcomes.
